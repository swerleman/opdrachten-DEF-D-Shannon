```python
import numpy as np

def capaciteit_plaatcondensator(oppervlakte, afstand, epsilon_r):
    
    epsilon_0 = 8.854e-12  # F/m 
    
    C = epsilon_0 * epsilon_r * (oppervlakte / afstand)
    
    return C


c = capaciteit_plaatcondensator(0.01, 0.00107, 2)
print(f"De capaciteit van de plaatcondensator is: {c:.2e} F")

def capaciteit_plaatcondensator_ampere(stroom, spanning, f):
    
    C = stroom / (spanning * 2 * np.pi * f) 
    
    return C

c_ampere = capaciteit_plaatcondensator_ampere((1.65/1000), 8.0, 100000)
print(f"Berekende capaciteit plaatcondensator groep 9 is {c_ampere:.2e} F")

epsilon_0 = 8.854e-12
```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.
import matplotlib.pyplot as plt
Minimale_spanning_osc = 190 #mV

g=9.81
u = np.array([0, 2, 9]) #kg
gemeten_capaciteit1 = np.array([23.6, 38.3, 101]) #pF

def c_functie(epsilon_r,delta_d,E_modulus,massa):
    return (epsilon_0*epsilon_r*g*massa)/(delta_d*E_modulus)*10e12

test_massa = np.linspace(0,9,1000)

plt.figure(figsize=(10, 6))
plt.plot(u, gemeten_capaciteit1, 'o-', label='Gemeten Capaciteit')
#spons is heel veel lucht, dus we hebben dezelfde epsilonr genomen. We hebben een kleine delta_d, om een nette lijn te kunnen maken. 
#De E is genomen van een spons, dus ongeveer 10 kPa.
plt.plot(test_massa,c_functie(1,0.005,8000,test_massa), label = 'Verwachtte Capaciteit')
plt.xlabel('massa (kg)')
plt.ylabel('Capaciteit (pF)')
plt.title('Kalibratiegrafiek van de Sensor')
plt.legend()
plt.grid()
plt.show()  

#Zoals te zien lijken de grafiek een beetje op elkaar, toch zijn ze niet gelijk. 
#Dit komt waarschijnlijk o.a. doordat de Elasticiteitsmodulus geschat is, en doordat we geen rekening gehouden hebben met een startende capaciteit.
```

```python
# Plot hier je oorspronkelijke en je uiteindelijke kalibratiegrafieken, fit daar een bijpassende lijn door
# en zet ook de verwachte ontwerpgrafiek erbij
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
Minimale_spanning_osc = 190 #mV

massa = np.array([0, 0.454, 0.600, 1.0 , 1.6, 2.6, 3.6, 4.1,6.2, 10]) #kg
gemeten_capaciteit = np.array([111.1, 113.0, 113.5, 114.9, 116.9, 119.0 ,123.7, 125.6 ,127.7 ,136.6]) #pF

u_massa = 0.5e-3 #weegschaal meet op gram
u_bits = np.array([4,3,5,2,3,4,2,1,4,5]) #gebaseerd op oscilaties van de bits

# physics-linearized model: Cbegin is massa 0  en a is de stijging van C per kg a = dC/dm
def model(m, Cbegin, a):
    return Cbegin + a*m

popt, pcov = curve_fit(model, massa, gemeten_capaciteit)
Cbegin, a = popt

print(f"C(m) = {Cbegin:.3f} + {a:.3f} * m   [pF]")
print("pcov:\n", pcov)

m_fit = np.linspace(massa.min(), massa.max(), 300)
C_fit = model(m_fit, *popt)

plt.figure(figsize=(10,6))
plt.errorbar(massa, gemeten_capaciteit, xerr=u_massa,yerr=u_bits, fmt='o', label='Gemeten data')
plt.plot(m_fit, C_fit, '-', label='Fit van condensator model')
plt.xlabel('Massa (kg)')
plt.ylabel('Capaciteit (pF)')
plt.title('Massa vs Capaciteit (lineair, fysisch onderbouwd)')
plt.grid(True)
plt.legend()
plt.show()


u = np.array([0, 2, 9]) #kg
gemeten_capaciteit1 = np.array([23.6, 38.3, 101]) #pF

def c_functie(epsilon_r,delta_d,E_modulus,massa):
    return (epsilon_0*epsilon_r*g*massa)/(delta_d*E_modulus)*10e12

test_massa = np.linspace(0,9,1000)

plt.figure(figsize=(10, 6))
plt.plot(u, gemeten_capaciteit1, 'o-', label='Gemeten Capaciteit')
plt.plot(test_massa,c_functie(1,0.005,8000,test_massa), label = 'Verwachtte Capaciteit')
plt.xlabel('massa (kg)')
plt.ylabel('Capaciteit (pF)')
plt.title('Kalibratiegrafiek van de Sensor')
plt.legend()
plt.grid()
plt.show()  



```

## Opdracht 10: Vul hieronder de conclusie in.

## Conclusie

Uit de metingen blijkt dat de capaciteit van de condensator lineair toeneemt met de massa die op de sensor wordt geplaatst. De lineaire fit van de meetdata geeft:

$C(m) = 112.53 + 2.54\,m \quad [pF]$

Dit betekent dat de sensor een gevoeligheid van ongeveer $2.54\,pF/kg$** heeft.

Als er massa op de spons wordt geplaatst worden de platen dichter naar elkaar gedrukt. Daardoor neemt de capaciteit van de condensator toe. Met deze formule is te zien wat de capaciteit is:

$C = \frac{\varepsilon A}{d}$

neemt de capaciteit toe wanneer de plaatafstand $d$ kleiner wordt.

Voor kleine vervormingen van de spons kan deze relatie worden benaderd als lineair, wat overeenkomt met de gemeten data.

Daarnaast laat de kalibratiegrafiek tussen spanning en capaciteit (gemeten met oscilloscoop) zien dat ook deze lijn ook lineair is. Dit betekent dat de sensor goed kan worden gebruikt om massa te bepalen.

De sensor kan worden gebruikt als een kracht sensor. De massa wordt door de elasticiteit van de spons wordt omgezet in een verandering van capaciteit.
